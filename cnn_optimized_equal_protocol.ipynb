```python
"""Colab: equally optimized CNN models, 5 folds x 10 epochs."""
import gc, os, random, shutil, subprocess
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import torch
import torch.nn as nn
from google.colab import drive
from IPython.display import display
from sklearn.metrics import (accuracy_score, confusion_matrix, f1_score,
    precision_score, recall_score, roc_auc_score, roc_curve)
from sklearn.model_selection import StratifiedKFold
from sklearn.utils.class_weight import compute_class_weight
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, models, transforms

DATA_DIR = "/content/breast_mri_dataset"
RAR_SOURCE = "/content/drive/MyDrive/data_set_zip/breast_mri_dataset.rar"
OUTPUT_ROOT = Path("/content/cnn_optimized_equal_protocol_outputs")
MODELS = ["ResNet34", "DenseNet121", "EfficientNet-B0", "MobileNetV2", "InceptionV3"]
IMG_SIZE, MICRO_BATCH, ACCUMULATION_STEPS = 224, 4, 8  # effective batch=32
EPOCHS, FOLDS, SEED, CLASSES = 10, 5, 42, 2
LR, WEIGHT_DECAY, LABEL_SMOOTHING, GRAD_CLIP = 1e-4, 1e-4, 0.1, 1.0

def seed_all(seed):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic=True; torch.backends.cudnn.benchmark=False

seed_all(SEED); device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
drive.mount("/content/drive")
shutil.copy2(RAR_SOURCE, "/content/breast_mri_dataset.rar")
subprocess.run(["unrar","x","-o+","/content/breast_mri_dataset.rar","/content/"], check=True)
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

mean=[.485,.456,.406]; std=[.229,.224,.225]
train_tf=transforms.Compose([transforms.Resize((IMG_SIZE,IMG_SIZE)),
    transforms.RandomRotation(15), transforms.RandomHorizontalFlip(.5),
    transforms.RandomVerticalFlip(.2), transforms.RandomAffine(0,translate=(.05,.05),scale=(.95,1.05)),
    transforms.ColorJitter(brightness=.15,contrast=.15), transforms.ToTensor(),
    transforms.Normalize(mean,std), transforms.RandomErasing(p=.15,scale=(.02,.08),ratio=(.3,3.3))])
eval_tf=transforms.Compose([transforms.Resize((IMG_SIZE,IMG_SIZE)),transforms.ToTensor(),transforms.Normalize(mean,std)])
train_aug=datasets.ImageFolder(os.path.join(DATA_DIR,"train"),transform=train_tf)
train_eval=datasets.ImageFolder(os.path.join(DATA_DIR,"train"),transform=eval_tf)
test_set=datasets.ImageFolder(os.path.join(DATA_DIR,"test"),transform=eval_tf)
assert train_aug.samples==train_eval.samples and train_eval.class_to_idx==test_set.class_to_idx
class_names=train_eval.classes; targets=np.asarray(train_eval.targets)
print("Device:",device,"| Classes:",train_eval.class_to_idx,"| Effective batch:",MICRO_BATCH*ACCUMULATION_STEPS)
test_loader=DataLoader(test_set,batch_size=MICRO_BATCH,shuffle=False,num_workers=2,pin_memory=torch.cuda.is_available())
splits=list(StratifiedKFold(FOLDS,shuffle=True,random_state=SEED).split(np.zeros(len(targets)),targets))
manifest=[]
for f,(ti,vi) in enumerate(splits,1):
    for split,idxs in (("Train",ti),("Validation",vi)):
        manifest += [{"Fold":f,"Split":split,"File":train_eval.samples[i][0],"Label":int(targets[i])} for i in idxs]
pd.DataFrame(manifest).to_csv(OUTPUT_ROOT/"Fold Assignments.csv",index=False)

def build(name):
    if name=="ResNet34":
        m=models.resnet34(weights=models.ResNet34_Weights.IMAGENET1K_V1); m.fc=nn.Sequential(nn.Dropout(.3),nn.Linear(m.fc.in_features,CLASSES))
    elif name=="DenseNet121":
        m=models.densenet121(weights=models.DenseNet121_Weights.IMAGENET1K_V1); m.classifier=nn.Sequential(nn.Dropout(.3),nn.Linear(m.classifier.in_features,CLASSES))
    elif name=="EfficientNet-B0":
        m=models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1); m.classifier=nn.Sequential(nn.Dropout(.3),nn.Linear(m.classifier[1].in_features,CLASSES))
    elif name=="MobileNetV2":
        m=models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1); m.classifier=nn.Sequential(nn.Dropout(.3),nn.Linear(m.classifier[1].in_features,CLASSES))
    elif name=="InceptionV3":
        m=models.inception_v3(weights=models.Inception_V3_Weights.IMAGENET1K_V1,aux_logits=True); m.aux_logits=False; m.AuxLogits=None; m.fc=nn.Sequential(nn.Dropout(.3),nn.Linear(m.fc.in_features,CLASSES))
    return m.to(device)  # all parameters remain trainable: full fine-tuning

def epoch_train(m,loader,opt,loss_fn,scaler):
    m.train(); opt.zero_grad(set_to_none=True); loss_sum=correct=total=0
    for step,(x,y) in enumerate(loader,1):
        x,y=x.to(device),y.to(device)
        with torch.autocast(device_type=device.type,dtype=torch.float16,enabled=device.type=="cuda"):
            out=m(x); raw_loss=loss_fn(out,y); loss=raw_loss/ACCUMULATION_STEPS
        scaler.scale(loss).backward()
        if step%ACCUMULATION_STEPS==0 or step==len(loader):
            scaler.unscale_(opt); nn.utils.clip_grad_norm_(m.parameters(),GRAD_CLIP)
            scaler.step(opt); scaler.update(); opt.zero_grad(set_to_none=True)
        loss_sum+=raw_loss.item()*y.size(0); correct+=(out.argmax(1)==y).sum().item(); total+=y.size(0)
    return loss_sum/total,correct/total

def loss_acc(m,loader,loss_fn):
    m.eval(); ls=ok=n=0
    with torch.inference_mode():
        for x,y in loader:
            x,y=x.to(device),y.to(device)
            with torch.autocast(device_type=device.type,dtype=torch.float16,enabled=device.type=="cuda"): out=m(x); loss=loss_fn(out,y)
            ls+=loss.item()*y.size(0); ok+=(out.argmax(1)==y).sum().item(); n+=y.size(0)
    return ls/n,ok/n

def predict(m,loader):
    m.eval(); yy=[]; pp=[]; pr=[]
    with torch.inference_mode():
        for x,y in loader:
            with torch.autocast(device_type=device.type,dtype=torch.float16,enabled=device.type=="cuda"): out=m(x.to(device))
            yy.extend(y.numpy()); pp.extend(out.argmax(1).cpu().numpy()); pr.extend(out.softmax(1)[:,1].cpu().numpy())
    return np.array(yy),np.array(pp),np.array(pr)

def scores(y,p,prob):
    try: auc=roc_auc_score(y,prob)
    except ValueError: auc=np.nan
    return {"Accuracy":accuracy_score(y,p),"Precision":precision_score(y,p,zero_division=0),
        "Recall":recall_score(y,p,zero_division=0),"F1-Score":f1_score(y,p,zero_division=0),"ROC-AUC":auc}

def learning_plots(h,name,folder):
    ep=range(1,len(h["Train Accuracy"])+1)
    for v in ("Accuracy","Loss"):
        plt.figure(figsize=(8,5)); plt.plot(ep,h[f"Train {v}"],"o-",label=f"Train {v}"); plt.plot(ep,h[f"Validation {v}"],"o-",label=f"Validation {v}")
        plt.title(f"{name} - {v}"); plt.xlabel("Epoch"); plt.ylabel(v); plt.legend(); plt.grid(alpha=.3)
        if v=="Accuracy": plt.ylim(0,1)
        plt.tight_layout(); plt.savefig(folder/f"{name} - {v}.png",dpi=300); plt.show(); plt.close()

def split_plots(y,p,prob,s,name,split,folder):
    plt.figure(figsize=(6,5)); sns.heatmap(confusion_matrix(y,p,labels=[0,1]),annot=True,fmt="d",cmap="Blues",xticklabels=class_names,yticklabels=class_names)
    plt.title(f"{name} - {split} Confusion Matrix"); plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.tight_layout(); plt.savefig(folder/f"{name} - {split} Confusion Matrix.png",dpi=300); plt.show(); plt.close()
    if not np.isnan(s["ROC-AUC"]):
        fpr,tpr,_=roc_curve(y,prob); plt.figure(figsize=(7,5)); plt.plot(fpr,tpr,label=f"AUC = {s['ROC-AUC']:.4f}"); plt.plot([0,1],[0,1],"--"); plt.legend(); plt.grid(alpha=.3)
        plt.title(f"{name} - {split} ROC Curve"); plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate"); plt.tight_layout(); plt.savefig(folder/f"{name} - {split} ROC Curve.png",dpi=300); plt.show(); plt.close()
    plt.figure(figsize=(8,5)); bars=plt.bar(s.keys(),s.values(),color="#1f77b4"); plt.ylim(0,1); plt.title(f"{name} - {split} Metrics")
    for b,v in zip(bars,s.values()): plt.text(b.get_x()+b.get_width()/2,min(v+.015,.98),f"{v:.4f}",ha="center")
    plt.tight_layout(); plt.savefig(folder/f"{name} - {split} Metrics.png",dpi=300); plt.show(); plt.close()

def run(name):
    root=OUTPUT_ROOT/name; root.mkdir(exist_ok=True); histories=[]; rows={"Train":[],"Validation":[],"Test":[]}
    for fold,(ti,vi) in enumerate(splits,1):
        seed_all(SEED+fold); gen=torch.Generator().manual_seed(SEED+fold); folder=root/f"Fold {fold}"; folder.mkdir(exist_ok=True)
        tr=DataLoader(Subset(train_aug,ti),MICRO_BATCH,shuffle=True,generator=gen,num_workers=2,pin_memory=torch.cuda.is_available())
        tre=DataLoader(Subset(train_eval,ti),MICRO_BATCH,shuffle=False,num_workers=2,pin_memory=torch.cuda.is_available())
        va=DataLoader(Subset(train_eval,vi),MICRO_BATCH,shuffle=False,num_workers=2,pin_memory=torch.cuda.is_available())
        weights=compute_class_weight("balanced",classes=np.unique(targets[ti]),y=targets[ti]); loss_fn=nn.CrossEntropyLoss(weight=torch.tensor(weights,dtype=torch.float32,device=device),label_smoothing=LABEL_SMOOTHING)
        m=build(name); opt=torch.optim.AdamW(m.parameters(),lr=LR,weight_decay=WEIGHT_DECAY); sched=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=EPOCHS,eta_min=1e-6)
        scaler=torch.amp.GradScaler("cuda",enabled=device.type=="cuda"); h={"Train Loss":[],"Train Accuracy":[],"Validation Loss":[],"Validation Accuracy":[]}; best=(-1,None,1)
        for ep in range(1,EPOCHS+1):
            tl,ta=epoch_train(m,tr,opt,loss_fn,scaler); vl,va_acc=loss_acc(m,va,loss_fn); sched.step()
            for k,v in (("Train Loss",tl),("Train Accuracy",ta),("Validation Loss",vl),("Validation Accuracy",va_acc)): h[k].append(v)
            histories.append({"Model":name,"Fold":fold,"Epoch":ep,"Learning Rate":opt.param_groups[0]["lr"],"Train Loss":tl,"Train Accuracy":ta,"Validation Loss":vl,"Validation Accuracy":va_acc})
            if va_acc>best[0]: best=(va_acc,{k:v.detach().cpu().clone() for k,v in m.state_dict().items()},ep)
            print(f"{name} | Fold {fold} | Epoch {ep}/{EPOCHS} | Train {ta:.4f} | Validation {va_acc:.4f}")
        m.load_state_dict(best[1]); learning_plots(h,name,folder)
        for split,loader in (("Train",tre),("Validation",va),("Test",test_loader)):
            y,p,prob=predict(m,loader); s=scores(y,p,prob); rows[split].append({"Model":name,"Fold":fold,"Best Epoch":best[2],**s}); split_plots(y,p,prob,s,name,split,folder)
        # Model weights are intentionally not saved; figures/tables stay GitHub-friendly.
        del m,opt,sched,scaler; gc.collect(); torch.cuda.empty_cache() if torch.cuda.is_available() else None
    summary=[]
    for split,data in rows.items():
        df=pd.DataFrame(data)
        for metric in ("Accuracy","Precision","Recall","F1-Score","ROC-AUC"): summary.append({"Model":name,"Split":split,"Metric":metric,"Mean":df[metric].mean(),"Standard Deviation":df[metric].std()})
    with pd.ExcelWriter(root/f"{name} - Results.xlsx",engine="openpyxl") as w:
        pd.DataFrame(histories).to_excel(w,sheet_name="Epoch History",index=False)
        for split,data in rows.items(): pd.DataFrame(data).to_excel(w,sheet_name=f"{split} Results",index=False)
        pd.DataFrame(summary).to_excel(w,sheet_name="Mean and Std",index=False)
    for split in rows: print(f"\n{name} - {split} Results"); display(pd.DataFrame(rows[split]))
    display(pd.DataFrame(summary)); return summary

combined=[]
for model in MODELS: combined.extend(run(model))
combined=pd.DataFrame(combined); combined.to_excel(OUTPUT_ROOT/"CNN - Combined Results.xlsx",index=False); display(combined)
print("Completed:",OUTPUT_ROOT)
```
